```python
import onnxruntime as ort
import librosa
import numpy as np
import os

MODEL_PATH = r".\tatvaani_android\app\src\main\assets\tatvanet.onnx"
AUDIO_FOLDER = r".\test_audio\fake"

session = ort.InferenceSession(MODEL_PATH)

for fname in os.listdir(AUDIO_FOLDER):
    if fname.endswith(('.wav', '.flac', '.mp3')):
        path = os.path.join(AUDIO_FOLDER, fname)
        audio, sr = librosa.load(path, sr=16000, mono=True, duration=2.0)
        
        # pad or trim to exactly 32000 samples (2 seconds at 16kHz)
        if len(audio) < 32000:
            audio = np.pad(audio, (0, 32000 - len(audio)))
        else:
            audio = audio[:32000]
        
        # build 9 spectrograms (3 types x 3 deltas)
        specs = []
        for n_mels in [64, 128, 128]:
            mel = librosa.feature.melspectrogram(y=audio, sr=sr, n_mels=n_mels, n_fft=512, hop_length=160)
            mel_db = librosa.power_to_db(mel, ref=np.max)
            mel_db = librosa.util.fix_length(mel_db, size=201, axis=1)
            mel_db = (mel_db - mel_db.mean()) / (mel_db.std() + 1e-6)
            if mel_db.shape[0] != 128:
                mel_db = np.resize(mel_db, (128, 201))
            specs.append(mel_db)
        
        # deltas
        for i in range(3):
            specs.append(librosa.feature.delta(specs[i]))
        for i in range(3):
            specs.append(librosa.feature.delta(specs[i], order=2))
        
        spec_tensor = np.stack(specs)[np.newaxis].astype(np.float32)  # (1, 9, 128, 201)
        audio_tensor = audio[np.newaxis].astype(np.float32)           # (1, 32000)
        
        outputs = session.run(None, {'audio': audio_tensor, 'spec_features': spec_tensor})
        probs = outputs[0][0]
        print(f"{fname} → {probs}")
```

```python
import scipy.special

labels = ['safe', 'caution', 'danger']

for fname in os.listdir(AUDIO_FOLDER):
    if fname.endswith(('.wav', '.flac', '.mp3')):
        path = os.path.join(AUDIO_FOLDER, fname)
        audio, sr = librosa.load(path, sr=16000, mono=True, duration=2.0)
        
        if len(audio) < 32000:
            audio = np.pad(audio, (0, 32000 - len(audio)))
        else:
            audio = audio[:32000]
        
        specs = []
        for n_mels in [64, 128, 128]:
            mel = librosa.feature.melspectrogram(y=audio, sr=sr, n_mels=n_mels, n_fft=512, hop_length=160)
            mel_db = librosa.power_to_db(mel, ref=np.max)
            mel_db = librosa.util.fix_length(mel_db, size=201, axis=1)
            mel_db = (mel_db - mel_db.mean()) / (mel_db.std() + 1e-6)
            if mel_db.shape[0] != 128:
                mel_db = np.resize(mel_db, (128, 201))
            specs.append(mel_db)
        
        for i in range(3):
            specs.append(librosa.feature.delta(specs[i]))
        for i in range(3):
            specs.append(librosa.feature.delta(specs[i], order=2))
        
        spec_tensor = np.stack(specs)[np.newaxis].astype(np.float32)
        audio_tensor = audio[np.newaxis].astype(np.float32)
        
        outputs = session.run(None, {'audio': audio_tensor, 'spec_features': spec_tensor})
        logits = outputs[0][0]
        probs = scipy.special.softmax(logits)
        pred = labels[np.argmax(probs)]
        
        print(f"{fname}")
        print(f"  safe={probs[0]:.2%}  caution={probs[1]:.2%}  danger={probs[2]:.2%}  → {pred.upper()}\n")
```
